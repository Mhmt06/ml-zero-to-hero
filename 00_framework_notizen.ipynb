# Modul 00 – 6-Schritte-Framework für ML-Projekte

## Grundlagen
**ML-Prozess:** 1. Datenerfassung → 2. Datenmodellierung → 3. Bereitstellung (Deployment)

**ML-Algorithmus vs. normales Programm:**
- Normales Programm: Man gibt die Anleitung (das Rezept) vor
- ML: Man gibt Zutaten + fertigen Kuchen vor → **der Algorithmus ermittelt die Anleitung selbst**

**Die 6 Schritte (iterativer Prozess, kein einmaliger Ablauf):**
1. Problemdefinition
2. Daten
3. Evaluierung
4. Merkmale (Features)
5. Modellierung
6. Experimente

---

## 1. Problemdefinition

### Arten des Lernens
| Art | Beschreibung | Beispiel |
|---|---|---|
| **Überwachtes Lernen** | Daten **und** Labels liegen vor. Der Algorithmus lernt, welche Muster zum Label führen; im Training werden falsche Vorhersagen korrigiert | Eingabe: Krankenakte → Ausgabe: Herzkrankheit ja/nein |
| **Unüberwachtes Lernen** | Daten **ohne** Labels. Der Algorithmus gruppiert, der Mensch benennt die Gruppen danach (Clustering) | Eingabe: Kunden → Ausgabe: Gruppen (Computer, Konsole, alte Spiele) |
| **Transferlernen** | Nutzt das Wissen eines bestehenden ML-Modells und passt es an das eigene Problem an. Spart Zeit und Kosten, da ML-Modelle teuer zu trainieren sind | Vorhandenes Textmodell + Kfz-Versicherungsansprüche (Daten) + Ergebnisse (Labels) → Klassifizierung „schuldhaft / nicht schuldhaft“ |
| **Bestärkendes Lernen**  | Ein Programm handelt in einer Umgebung und lernt über Belohnung und Bestrafung | Programm lernt Schach zu spielen |

### Problemarten
| Art | Beschreibung | Beispiel |
|---|---|---|
| **Klassifizierung** | Kategorie vorhersagen; binär (2 Klassen) oder Mehrklassen | Kunde wandert ab: ja/nein |
| **Regression** | Eine bestimmte Zahl vorhersagen | Hauspreis |
| **Empfehlung** | Passende Produkte vorschlagen | Produktempfehlung anhand des Kaufverlaufs |

---

## 2. Daten

| Art | Beschreibung |
|---|---|
| **Strukturiert** | Tabellen, z. B. Patientendaten: Geschlecht, Herzfrequenz … |
| **Unstrukturiert** | Bilder, Audiodateien, Text … |
| **Statisch** | Vorhandene, historische Daten, ändern sich nicht |
| **Streaming** | Werden ständig aktualisiert |

- **Überwachtes Lernen:** Merkmalsvariablen (z. B. Gewicht) → sagen vorher → Zielvariable (z. B. Herzkrankheit)
- **Unüberwachtes Lernen:** keine Labels
- **Transferlernen:** ist meist ein **überwachtes** Problem, man nutzt aber Muster, die ML-Algorithmen aus anderen Datenquellen gelernt haben
- ⚠️ Fachexperten immer über die Daten informieren und einbeziehen

## 3. Evaluierung
Vorher festlegen, was Erfolg bedeutet, z. B. 95 % Genauigkeit.

### Bei Klassifizierungsproblemen
| Begriff | Bedeutung | Beispiel |
|---|---|---|
| **Falsch-negativ** | Modell sagt negativ, ist aber positiv | Autonomes Auto erkennt keinen Fußgänger, obwohl einer da ist |
| **Falsch-positiv** | Modell sagt positiv, ist aber negativ | Modell sagt Herzkrankheit, Person ist aber gesund |
| **Richtig-negativ** | Modell sagt negativ, ist negativ | |
| **Richtig-positiv** | Modell sagt positiv, ist positiv | |

| Metrik | Bedeutung |
|---|---|
| **Präzision (Precision)** | Anteil der positiven Vorhersagen, die korrekt sind. 1,0 = keine Falsch-positiven |
| **Recall** | Anteil der tatsächlich Positiven, die korrekt vorhergesagt wurden. 1,0 = keine Falsch-negativen |
| **F1-Score** | Kombination aus Präzision und Recall |
| **ROC-Kurve** | Vergleicht Trefferquote (Richtig-positiv-Rate) mit Falsch-positiv-Rate |
| **AUC** | Fläche unter der ROC-Kurve: 0,0 = 100 % falsch · 0,5 = Zufall · 1,0 = perfekt |

### Bei Regressionsproblemen
- **MAE** (Mean Absolute Error): mittlerer absoluter Fehler
- **RMSE** (Root Mean Squared Error): **Wurzel** aus dem mittleren quadratischen Fehler, bestraft große Fehler stärker

### Bei Empfehlungsproblemen
- Schwer zu testen, am besten durch Experimente
- **Precision@K:** Genauigkeit der ersten K Ergebnisse, z. B. aus 10.000 Produkten die 5 besten Empfehlungen

## 4. Merkmale (Features)

| Art | Beispiel |
|---|---|
| **Kategorisch** | Kauf getätigt (ja/nein), Geschlecht des Patienten |
| **Kontinuierlich (numerisch)** | Anzahl der Anmeldungen |

- **Feature Engineering:** aus vorhandenen Daten neue Merkmale bauen, z. B. Anmeldungen + Zeitstempel → „Zeit seit der letzten Anmeldung“
- Trainings- und Testdaten müssen ähnliche Merkmale haben
- Fachexperten kennen das Problem und wissen, welche Merkmale wichtig sind

---

## 5. Modellierung

### a) Auswahl eines Modells
Kriterien: Interpretierbarkeit und einfache Fehlersuche · Datenmenge · Einschränkungen bei Training und Vorhersage

- **Mit einfachen Lösungen beginnen:** lineare Modelle, logistische Regression
- Die reale Welt ist oft nicht linear → **Entscheidungsbäume und Gradient-Boosting-Algorithmen** (Random Forest, XGBoost, CatBoost) → gut für **strukturierte** Daten
- **Tiefe Modelle (neuronale Netze)** → gut für **unstrukturierte** Daten (Bilder, Audio)
  - ⚠️ Längere Trainingsdauer, schwierigere Fehlersuche, Vorhersage dauert länger
- **Transferlernen:** Kombination aus Deep Learning und linearen Modellen. Ein vortrainiertes Deep-Learning-Modell erkennt die Muster, ein lineares Modell nutzt sie als Eingabe
  - Wo finde ich vortrainierte Modelle? PyTorch Hub, TensorFlow Hub, Model Zoo, fast.ai

### b) Optimierung und Verbesserung
- Hyperparameter anpassen: Lernrate, Optimierer, Anzahl der Bäume beim Random Forest …
- Reproduzierbarkeit und Effizienz haben hohe Priorität

### c) Vergleich von Modellen
Nur fair, wenn alle Modelle auf **denselben Daten X trainiert** und auf **denselben Daten Y bewertet** werden

## 6. Experimente
→ Praxisrelevanz

- **Offline-Experimente:** noch nicht kundenorientiert
- **Online-Experimente:** im Produktivbetrieb

### Datenaufteilung
| Datensatz | Anteil | Schlechtes Ergebnis bedeutet |
|---|---|---|
| **Trainingsdaten** | 70–80 % | Modell hat nicht richtig gelernt |
| **Validierungs- / Entwicklungsdaten** | 10–15 % | |
| **Testdaten** | 10–15 % | Modell generalisiert nicht gut |

- Schlechtes Ergebnis in der Praxis → Trainings- und Testdaten weichen von der Realität ab

> **Immer daran denken:** Erst mit einem Machbarkeitsnachweis (Proof of Concept) prüfen, ob die Umsetzung funktioniert.
